```python
import numpy as np
import pandas as pd
df = pd.DataFrame({
    "Name":["Asha", "Ravi", "Maya", "John", "Banu", "Guna", "Tina"],
    "Marks":[80, 90, None, 75, 70, 85, 60],
    "City":["Hyderabad", "Delhi", "Hyderabad", "Mumbai", "Delhi", None, "Mumbai"]
})
```

### Grouping and Aggregation

```python
df.groupby("City")["Marks"].mean()
```

```python
df.groupby("City")["Marks"].agg(["mean","min", "max", "count", "size"])
```

```python
df.groupby("City").agg(
    Avg_marks=("Marks","mean"), 
    Lowest_marks=("Marks","min"), 
    Highest_marks=("Marks","max"), Attempted=("Marks","count"))
```

### Merging DataFrames

```python
student = pd.DataFrame({
   "ID": ["1", "2", "3", "4", "5", "6", "7"],
    "Name":["Asha", "Banu", "Guna", "John", "Maya", "Ravi", "Tina"]})
```

```python
score = pd.DataFrame({
     "ID": ["1", "2", "3", "4", "5", "6", "7"],
    "Score": [80, 70, 85, 75, 78, 90, 60]
})
```

```python
pd.merge(student, score, on="ID", how = "inner")
```
